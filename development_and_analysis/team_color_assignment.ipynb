```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
```

```python
img_path = "../output/cropped_img.jpg"
img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
plt.show()
```

```python
top_half_img = img[0: int(img.shape[0] / 2), :]
plt.imshow(top_half_img)
plt.show()
```

```python
img_2d = top_half_img.reshape(-1, 3)

kmeans = KMeans(n_clusters=2, random_state=0).fit(img_2d)

labels = kmeans.labels_

clustered_img = labels.reshape(top_half_img.shape[0], top_half_img.shape[1])

plt.imshow(clustered_img)
plt.show()
```

```python
cluster_corners = [clustered_img[0,0], clustered_img[0, -1], clustered_img[-1, 0], clustered_img[-1, -1]]
cluster_non_player = max(set(cluster_corners), key=cluster_corners.count)
print(cluster_non_player_)
```

```python
cluster_player = 1 - cluster_non_player
print(cluster_player)
```

```python
kmeans.cluster_centers_[cluster_player]
```
